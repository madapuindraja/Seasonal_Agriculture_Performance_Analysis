```python
df["Rainfall_mm"] = df["Rainfall_mm"].fillna(df["Rainfall_mm"].median())

df["Soil_Moisture_pct"] = df["Soil_Moisture_pct"].fillna(df["Soil_Moisture_pct"].median())

df["Yield_Tonnes_Ha"] = df["Yield_Tonnes_Ha"].fillna(df["Yield_Tonnes_Ha"].median())
```

```python
season_performance = df.groupby("Season")[[
    "Yield_Tonnes_Ha",
    "Production_Tonnes",
    "Revenue_INR",
    "Total_Cost_INR",
    "Profit_INR",
    "Water_Used_m3",
    "Water_Efficiency_t_per_1000m3",
    "Disease_Pest_Risk_pct"
]].mean()

season_performance
```

```python
best_season = season_performance["Profit_INR"].idxmax()
best_profit = season_performance["Profit_INR"].max()

print("Best season:", best_season)
print("Average profit:", round(best_profit, 2))
```

```python
profit_comparison = season_performance["Profit_INR"].sort_values(ascending=False)

profit_comparison
```

```python
profit_difference = (
    season_performance["Profit_INR"]["Kharif"]
    - season_performance["Profit_INR"]["Rabi"]
)

print("Kharif vs Rabi profit difference: ₹", round(profit_difference, 2))
```

```python
profit_difference_zaid = (
    season_performance["Profit_INR"]["Kharif"]
    - season_performance["Profit_INR"]["Zaid"]
)

print("Kharif vs Zaid profit difference: ₹", round(profit_difference_zaid, 2))
```

```python
id="p8k2lm"
zaid_profit = season_performance.loc["Zaid", "Profit_INR"]
kharif_profit = season_performance.loc["Kharif", "Profit_INR"]

percentage_difference = (
    (kharif_profit - zaid_profit) / abs(zaid_profit)
) * 100

print("Kharif profit difference compared with Zaid:",
      round(percentage_difference, 2), "%")
```
